```python
import pandas as pd

# Crear un DataFrame con las deudas
deudas = pd.DataFrame({
    "Deuda": [f"Deuda {i}" for i in range(1, 11)],
    "Monto": [20000, 15000, 10000, 8000, 4000, 4000, 11000, 35000, 5000, 30000]
})

# Inicializar variables
ingreso_mensual = 10000
pagos_mensuales = 8000  # 80% del ingreso
plan = []

# Lista para registrar deudas activas
deudas_act = deudas.copy()
deudas_act["Pagado"] = 0

meses = 10
for mes in range(1, meses + 1):
    pago_restante = pagos_mensuales
    mes_actual = f"Mes {mes}"

    for idx in deudas_act.index:
        deuda_nombre = deudas_act.loc[idx, "Deuda"]
        deuda_monto = deudas_act.loc[idx, "Monto"]
        pagado = deudas_act.loc[idx, "Pagado"]
        restante = deuda_monto - pagado

        if restante == 0:
            continue

        pago = min(restante, pago_restante)
        if pago > 0:
            plan.append({
                "Mes": mes_actual,
                "Deuda": deuda_nombre,
                "Pago": pago,
                "Restante después de pago": restante - pago
            })
            deudas_act.loc[idx, "Pagado"] += pago
            pago_restante -= pago

        if pago_restante == 0:
            break

# Crear DataFrame del plan
plan_df = pd.DataFrame(plan)

# Total de pagos por mes
resumen_mensual = plan_df.groupby("Mes")["Pago"].sum().reset_index()
resumen_mensual.columns = ["Mes", "Total Pagado"]

# Guardar a Excel
with pd.ExcelWriter("Plan_Pago_Deudas.xlsx") as writer:
    plan_df.to_excel(writer, sheet_name="Plan Detallado", index=False)
    resumen_mensual.to_excel(writer, sheet_name="Resumen Mensual", index=False)

print("Archivo 'Plan_Pago_Deudas.xlsx' creado con éxito.")
```
